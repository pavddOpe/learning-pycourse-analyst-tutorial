## **Challenge**

Create a program that receives a list of numbers as input and prints a new list that:

- Contains the original list followed by its reverse
- Has the first element of the original list inserted at the beginning and the last element inserted at the end
- Repeats this entire sequence twice

```python
numbers = input().split()
new_list = (numbers[:1] + (numbers+numbers[::-1]) + numbers[-1:])*2
print(new_list)
```

```python
# the programm for CODDY
print('Welcome to the Daily Expense Tracker!')
print(f'\nMenu:\n1. Add a new expense\n2. View all expenses\n3. Calculate total and average expense\n4. Clear all expenses\n5. Exit')
expense_lst=[]
expenses_total = 0
while True:  #infinite while loop
    choice = input()
    if choice == '1':
        while True:
            expense=float(input())
            if expense > 0:
                expense_lst.append(expense)
                expenses_total+=1
            print('Expense added successfully!')
            break
    if choice == '2':
        if not expense_lst:
            print('No expenses recorded yet.')
            continue
        else:
            print('Your expenses:')
            for i,value in enumerate(expense_lst):
                print(f'{i+1}. {value}')
            continue
        break
    if choice == '3':
        if not expense_lst:
            print('No expenses recorded yet.')
            continue
        else:
            print('Total expense:', sum(expense_lst))
            average=sum(expense_lst)/expenses_total
            print('Average expense:', average)
            continue
    if choice == '4':
        expense_lst.clear()
        expenses_total-=expenses_total
        print('All expenses cleared.')
        continue
    if choice == '5':
        print('Exiting the Daily Expense Tracker. Goodbye!')
        break
    if choice not in ['1','2','3','4','5']:
        print('Invalid choice. Please try again.')
    continue
```
