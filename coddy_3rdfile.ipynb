```python
# the programm for CODDY
print('Welcome to the Daily Expense Tracker!')
print(f'\nMenu:\n1. Add a new expense\n2. View all expenses\n3. Calculate total and average expense\n4. Clear all expenses\n5. Exit')
expense_lst=[]
expenses_total = 0
while True:  #infinite while loop
    choice = input()
    if choice == '1':
        while True:
            expense=float(input())
            if expense > 0:
                expense_lst.append(expense)
                expenses_total+=1
            print('Expense added successfully!')
            break
    if choice == '2':
        if not expense_lst:
            print('No expenses recorded yet.')
            continue
        else:
            print('Your expenses:')
            for i,value in enumerate(expense_lst):
                print(f'{i+1}. {value}')
            continue
        break
    if choice == '3':
        if not expense_lst:
            print('No expenses recorded yet.')
            continue
        else:
            print('Total expense:', sum(expense_lst))
            average=sum(expense_lst)/expenses_total
            print('Average expense:', average)
            continue
    if choice == '4':
        expense_lst.clear()
        expenses_total-=expenses_total
        print('All expenses cleared.')
        continue
    if choice == '5':
        print('Exiting the Daily Expense Tracker. Goodbye!')
        break
    if choice not in ['1','2','3','4','5']:
        print('Invalid choice. Please try again.')
    continue
```

- Takes two string arguments: text and pattern
- Counts how many times pattern appears in text, including overlapping occurrences
- Returns a tuple containing:
- A boolean indicating if the pattern was found (True/False)
- The number of occurrences of the pattern
- A list of starting positions where the pattern was found

For example, if text is "abababab" and pattern is "aba", your function should return (True, 3, [0, 2, 4]), since "aba" appears at positions 0, 2, and 4.

If the pattern is not found, return (False, 0, []).

**About the pass keyword**:<br> You'll see pass in the default code. It's a Python keyword that means "do nothing" and is used as a placeholder when Python requires an indented code block (like inside a function). You should replace pass with your actual function code.<br>
**Pass** *means "do nothing" and is used as a placeholder when Python requires an indented code block (like inside a function)*

```python
def find_occurrences(text, pattern):
    lst_positions=[]                    #or use count = 0 here
    for i in range(len(text)):
        if text[i:i+len(pattern)] == pattern:
            lst_positions.append(i)            
    if lst_positions:
        return (True,len(lst_positions),lst_positions)      #and then count inside this tuple
    else:
        return '(False, 0, [])'
    
# Read input
text = input()
pattern = input()

# Call your function and print the result
result = find_occurrences(text, pattern)
print(result)
```

**Write a program that receives three inputs (given):**

- A list of prices
- A list of item names
- A budget per item<br> 

**The program should print:**
- A list of items that you can afford within your budget
- How much budget would you need if you bought all of the affordable items
- How many items you couldn't afford'

```python
prices = input().split(",")
for i in range(len(prices)):
    prices[i]=int(prices[i])
items = input().split(",")
budget_per_item = int(input())

affordable_items = []
cant_afford = 0
total_needed = 0


# Write your code below
for item in range(len(items)):
    product = items[item]
    price = prices[item]
    if price <= budget_per_item:
        affordable_items.append(product)
        total_needed+=price
    else:
        cant_afford+=1
print("Can buy:", affordable_items)
print("Total budget needed:", total_needed)
print("Can't afford:", cant_afford)
```

**Create a function named calculate_discount** that takes two parameters:

 - price: The original price of an item (float)
- discount_percentage: The discount percentage (float)<br>

**The function should:**
- Calculate the discount amount
- Subtract the discount amount from the original price
- Round the result to 2 decimal places
- Return the final discounted price

```python
price = float(input())
discount_percentage = float(input())
def calculate_discount(price, discount_percentage):
    discount_amount = price * (discount_percentage/100)
    result = price - discount_amount
    return round(result,2)
calculate_discount(price,discount_percentage)
```

**Create a function named update_inventory** that takes three parameters:<br>inventory_dict (a dictionary), item (a string), and quantity (an integer).<br> The function should update the inventory_dict with the new item and quantity.<br>If the item already exists in the inventory, its quantity should be increased by the given amount. If the item does not exist, a new item should be added with the given quantity.<br> The function should return the updated dictionary.

```python
def update_inventory(inventory_dict, item, quantity):
    if item in inventory_dict:
        inventory_dict[item]+=quantity
    else:
        inventory_dict[item]=quantity
    return inventory_dict
update_inventory({"sugar":12,"coffee":7},'tea',5)
```

#### **Dictionary methods**
- **keys()** - my_dict.keys() - *Returns a view object that displays a list of all the keys in the dictionary*
- **values()** - my_dict.values() - *Returns a view object that displays a list of all the values in the dictionary*
- **items()** - my dict.items() - *Returns a view object that displays a list of a dictionary's key-value tuple pairs*
- **get(key, default)** - my_dict.get(key,default) - *Returns the value for the specified key. If the key is not found, it returns the default value (or None if no default is specified)*
- **pop(key)** - my_dict.pop(key) - *Removes the element with the specified key and returns its value*

```python
# Step 1: Create the Grades Dictionary
grades = {
   'Alice':85,'Bob':90,'Charlie':78
}

# Step 2: Access All Keys and Values
# Print all students and grades
print(f'Students:',grades.keys())
print(f'Grades:',grades.values())
# Step 3: Add a New Student
# Add Diana with a grade of 92
grades['Diana']=92
# Step 4: Retrieve a Student's Grade
# Get Bob's grade using get() method
print(f"Bob's grade:",grades.get('Bob'))
# Step 5: Remove a Student
# Remove Charlie using pop() method
# Print the updated dictionary
grades.pop('Charlie')
print(f'Updated grades:',grades)
```
