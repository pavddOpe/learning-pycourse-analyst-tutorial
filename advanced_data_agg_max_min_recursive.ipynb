**Write a function setup_coastal_study** that takes books, lamp_level, materials and returns a dictionary with the organized study space setup.
The function organizes your coastal study area by reversing book titles for better shelf arrangement, adjusting lamp brightness, and preparing materials.

**Logic:**
- Reverse every other book title starting from the second book (index 1, 3, 5, etc.)
- Increment lamp level by 2, then decrement by 1 to reach optimal brightness
- Reverse the entire materials list for better accessibility

Parameters:
- books (list): List of book titles to organize
- lamp_level (int): Starting brightness level
- materials (list): List of study materials<br>
**Returns:** Dictionary with organized study setup. Format: {"books": [...], "lamp_level": int, "materials": [...]}

```python
def setup_coastal_study(books,lamp_level,materials):
    processed_books=[]
    for b in range(len(books)):
        if b % 2 == 1:
            b_rev = books[b][::-1]
            processed_books.append(b_rev)
        else:
            processed_books.append(books[b])
    lamp_level = lamp_level + 2
    lamp_level = lamp_level - 1
    rev_mat=materials[::-1]
    fin_dict = {
        'books': processed_books,
        'lamp_level': lamp_level,
        'materials': rev_mat
    }
    print(fin_dict)
setup_coastal_study(['Coastal Erosion', 'Salt Water', 'Marine Life', 'Ocean Chemistry', 'Seaweed Studies', 'Fish Behavior'],10,['test tubes', 'pH strips', 'magnifier'])
```

**Create a function named analyze_grades** that takes a dictionary of student grades as an argument. The dictionary keys are student names, and the values are their corresponding grades. The function should perform the following operations:

**Calculate the average grade of all students.**
- Find the highest and lowest grades.
- Identify the student(s) with the highest and lowest grades.
- Return a dictionary containing the following information:
- 'average': The average grade (rounded to 2 decimal places)
- 'highest': The highest grade
- 'lowest': The lowest grade
- 'top_student': The name of the student with the highest grade
- 'bottom_student': The name of the student with the lowest grade

```python
def analyze_grades(grades):
    avg=sum(grades.values())/len(grades.values())
    highest = max(grades.values())
    lowest = min(grades.values())
    for student,grade in grades.items():
        if grade == highest:
            top_student=student
        elif grade == lowest:
            bottom_student=student
    
    fin_dic={
        'average': avg,
        'highest': highest,
        'lowest': lowest,
        'top_student': top_student,
        'bottom_student': bottom_student
    }
    return fin_dic
# Test the function
student_grades = {'Alice': 85, 'Bob': 92, 'Charlie': 78, 'David': 95, 'Eve': 88}
result = analyze_grades(student_grades)
print(result)
```

The function helps you decide whether to keep or discard your messy roommate's belongings based on their condition and type.<br>

**Conditions:**
- Keep if condition is "clean"
- Keep if item_type is "electronics" and condition is not "broken"
- Keep if item_type is "book" and condition is not "moldy"
- Discard everything else

**Parameters:**
- item_name (str): Name of the item
- condition (str): Current condition ("clean", "dirty", "stained", "broken", "moldy")
- item_type (str): Type of item ("clothing", "electronics", "book")<br>
**Returns**: Dictionary with decision and reason. Format: {"decision": "keep", "reason": "item is clean"}

```python
def sort_roommate_item(item_name,condition,item_type):
    fin_dict={}
    if condition == "clean":
        fin_dict['decision']='keep'
        fin_dict['reason']='item is clean'
    elif item_type == "electronics" and not condition == "broken":
        fin_dict['decision']='keep'
        fin_dict['reason']='electronics in working condition'
    elif item_type == "book" and not condition == "moldy":
        fin_dict['decision']='keep'
        fin_dict['reason']='book is readable'
    else:
        fin_dict['decision']='discard'
        fin_dict['reason']='item does not meet keep criteria'

    return fin_dict
sort_roommate_item('tablet','moldy','electronics')
    
```

**Write a function manage_stand** that takes inventory, orders and returns a dictionary summarizing remaining fresh vs. preserved items.

Your farm stand tracks inventory with preservative status and processes customer orders throughout the day.<br>
**Logic:**
- Parse each inventory item from "item:count:preservative" format
- Process each order to reduce stock for matching items
- Group remaining items by preservative status (fresh vs. artificial)
- Calculate totals for each category

**Parameters:**
- inventory (list): Initial stock in "item:count:preservative" format
- orders (list): Customer orders in "item:quantity" format<br>
**Returns**: Dictionary with preservative categories and their item counts. Format: {"fresh": {"item1": count1, "item2": count2}, "artificial": {"item3": count3}}

```python
def manage_stand(inventory,orders):
    stock={}
    preservative={}
    for item_str in inventory:
        parts = item_str.split(':')
        item = parts[0]
        count = int(parts[1])
        status = parts[2]

        stock[item]=count
        preservative[item]=status
    
    for order in orders:
        parts = order.split(':')
        item = parts[0]
        qty = int(parts[1])

        if item in stock:
            stock[item]-=qty
            if stock[item] < 0:
                stock[item] = 0

    result = {
        'fresh': {},
        'artificial': {}
    }
    for item, count in stock.items():
        if count>0:
            status=preservative[item]
            if status=='fresh':
                result['fresh'][item]=count
            elif status=='artificial':
                result['artificial'][item]=count
    print(result)
manage_stand(['strawberries:12:fresh', 'blueberries:8:fresh', 'raspberries:15:artificial'],['strawberries:4', 'blueberries:8', 'raspberries:7'])
manage_stand(['apple:5:fresh'],['apple:2'])
```

**Create a function named house_of_lists that takes a list of lists list_of_lists** as an argument. Each inner list contains numbers. The function should use list comprehensions to perform the following operations:

- Filter out inner lists that have a sum greater than 50, keeping only those whose sum is 50 or less.
- From the remaining inner lists, extract numbers that are less than 5.
- Combine all extracted numbers into a single list.<br>
**Return the combined list of numbers.**

```python
def house_of_lists(list_of_lists):
    return [number for sublist in list_of_lists
            if sum(sublist)<=50
            for number in sublist
            if number<5
            ]
house_of_lists([[5, 5, 5, 5, 5], [1, 1, 1, 1], [2, 2, 2, 2], [50, 1]])
```

**Create a function named elements_of_freedom** that processes a list of string elements according to specific rules.<br>

**Your function should:**
- Filter out elements that have fewer than 5 characters
- Convert the remaining elements to uppercase
- Remove any duplicate elements
- Return a list of the unique uppercase elements
- Use list operations to efficiently process the data.<br>

**Note**: Preserve the original order of elements in the list.

```python
def elements_of_freedom(elements):
    l1=[el for el in elements if len(el)>=5]
    l1=[el.upper() for el in l1]
    return list(dict.fromkeys(l1))
elements_of_freedom(["apple", "banana", "cherry", "date", "apple", "banana", "grape", "fig"])
```

A **recursive function** is a function that calls itself to solve smaller instances of a problem. Each recursive call must bring the function closer to a base case, which stops the recursion.

```python
def count_down(n):
    if n >= 0:
        print(n)
        count_down(n-1)
count_down(3)
```

Write a recursive function named **print_sequence** that takes a positive integer n as an argument and prints a countdown sequence. For each number from n to 1, it should print 'T-minus {number}'. When it reaches 0, it should print 'Blast off!'.

```python
def print_sequence(n):
    if n>0:
        print(f'T-minus {n}')
        print_sequence(n-1)
    elif n==0:
        print('Blast off!')
print_sequence(10)
```

Write a recursive function named fibonacci that takes a positive integer n as an argument and returns the nth Fibonacci number. 

```python
def fibonacci(n):
    if n == 1:
        return 0
    elif n == 2:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)
fibonacci(4)
```

Write a recursive **function named sum_digits** that takes a positive integer n as an argument and returns the sum of its digits. The function should work as follows:

If n is a single digit (less than 10), return that digit.
Otherwise, return the sum of the last digit of n and the result of sum_digits called with n divided by 10 (integer division).

```python
def sum_digits(n):
    if n < 10:
        return n
    else:
        return (n%10) + sum_digits(n//10)
sum_digits(11)
```

Write a **recursive function named sum_nested** that takes a nested list nested_list as an argument. The function should:

Add all the integers in the list, including integers in any sublists.
Return the total sum as an integer.

```python
def sum_nested(nested_list):
    list_1=[]
    for i in nested_list:
        if i is int:
            list_1.append(i)
    total=sum(list_1)
    return total
sum_nested([1, [2, 3], [4, [5, 6]], 7])
```
