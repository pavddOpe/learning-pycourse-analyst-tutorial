- Create a function called handle_shopping_cart that processes customer orders

**Process each order by:**

- Converting the quantity to an integer

3. Return the completed shopping cart dictionary

```python
def handle_shopping_cart(orders):
    # Create an empty dictionary for the shopping cart
    shop_cart={}
    # Process each order in the list
    for order in orders:
        part = order.split(':')
        if len(part) != 2:
            print(f'Invalid format: {order}')
            continue
        item = part[0]
        quantity_str=part[1]
        try:
            quantity = int(quantity_str)
            # Handle potential errors
        except ValueError:
            # Handle value errors
            print(f'Invalid quantity: {order}')
            continue
        if quantity < 0:
            print(f'Negative quantity not allowed: {order}')
            continue
        shop_cart[item]=shop_cart.get(item,0)+quantity
    return shop_cart
    # Return the completed cart
handle_shopping_cart(["chocolate:-5","yogurt:3","milk:2"])
```

**Create a function named get_prime_numbers** that takes a list of integers numbers as an argument. The function should use the filter() function to select prime numbers from the list. A prime number is a number greater than 1 that has no positive divisors other than 1 and itself. The function should return a list containing the selected prime numbers.

```python
def get_prime_numbers(numbers):
    # Define a helper function to check if a number is prime
    # A number n is prime if n >= 2 and it has no divisors between 2 and int(n**0.5) + 1
    def is_prime(n):
        # TODO: Return False if n is less than 2
        if n < 2:
            return False
        # TODO: Loop from 2 to int(n**0.5) + 1 and return False if any divisor divides n evenly
        else:
            for x in range(2,int(n**0.5)+1):
                if n % x == 0:
                    return False
            return True
        # TODO: Return True if no divisors were found

    # Use filter() with the helper function to select prime numbers
    prime_numbers = filter(is_prime, numbers)
    
    # Return the list of selected prime numbers
    return list(prime_numbers)
get_prime_numbers([47,48,49,50,51,53,59,60,61,67])
```

**Create a function named clean_email_list** that takes a list of email addresses and returns a list of valid, standardized email addresses.<br>

**Requirements:**
- Convert all emails to lowercase and strip all whitespace at the beginning or end
- Only keep emails that:
- Contain exactly one '@' symbol
- Have at least one character before the '@'
- Have at least one character after the ‘@’

Use map() and filter() to solve the problem.

```python
def clean_email_list(emails):
    def clean_email(email):
        return email.lower().strip(' ')
    def is_valid(email):
        parts=email.split('@')
        if len(parts) == 2 and len(parts[0]) >=1 and len(parts[1])>=1:
            return True
        return False
    cleaned=map(clean_email,emails)
    valid_emails=filter(is_valid,cleaned)
    return list(valid_emails)
clean_email_list(['USER@DOMAIN.COM',  ' space@domain.com ',  'NoSpaces@domain.com'])
```

**Create function named process_numbers** that takes a list of numbers and processes them according to specific rules.<br>

**Requirements:**
- First filter out all negative numbers and zero
- Then for the remaining positive numbers:
- Double even numbers
- Triple odd numbers<br>
Use map() and filter() to solve the problem.

```python
def process_numbers(numbers):
    def positive(number):
        if number > 0:
            return True
    def processing(number):
        if number % 2 == 0:
            return number * 2
        else:
            return number * 3
        
    positives=filter(positive,numbers)
    processed_numbers = map(processing,positives)
    return list(processed_numbers)
process_numbers([1, 2, 3, 4])
```

**Create a function named organize_contacts** that processes a list of contact dictionaries to create a clean contact database.<br>

**Each contact dictionary in the input list has these keys:**
- name: The person's name
- email: The person's email address
- phone: The person's phone number<br>

**Your function should:**
- Remove duplicate contacts (contacts with the same email or phone number)
- Standardize all emails to lowercase
- Filter out contacts with invalid email addresses
- Filter out contacts with invalid phone numbers
- Return a list of cleaned contact dictionaries
**Validation rules:**
- Valid email: Must contain '@' and '.', and must not have spaces
- Valid phone: Must contain exactly 10 digits (ignore non-digit characters like dashes or parentheses)
- For cleaning phone numbers, you should use Python's str.isdigit() method to extract only the numeric digits from phone numbers. This method returns True if a character is a digit (0-9) and False otherwise, making it perfect for filtering out non-digit characters.

```python
def organize_contacts(contact_list):
    if not contact_list:
        return []
    
    cleaned = []
    seen = set()

    for contact in contact_list:
        cont=contact.copy()

        if 'email' in cont and cont['email']:
            cont['email']=cont['email'].lower().strip()

        if 'phone' in cont and cont['phone']:
            digits=''.join(c for c in str(cont['phone']) if c.isdigit())
            cont['phone']=digits

        email=cont.get('email','')
        if not (email and '@' in email and '.' in email and ' ' not in email):
            continue

        phone = cont.get('phone','')
        if not (phone and len(phone)==10):
            continue

        email_key=cont.get('email')
        phone_key=cont.get('phone')

        if email_key in seen or phone_key in seen:
            continue

        if email_key:
            seen.add(email_key)
        if phone_key:
            seen.add(phone_key)
        
        cleaned.append(cont)
    return cleaned
organize_contacts([{"name": "John Doe", "email": "JOHN@email.com", "phone": "123-456-7890"}, {"name": "Jane Doe", "email": "jane@email.com", "phone": "123.456.7890"}, {"name": "Bob Smith", "email": "invalid.email", "phone": "12345"}])
```

Create a **function named analyze_text** that analyzes words in a text string. HARD<br>

1. Count unique words (case-insensitive)
2. Find all words that appear more than once
3. Identify all palindrome words (words that read the same forwards and backwards, like "level")
4. Return a dictionary with three keys:
- unique_count: the number of unique words (count also repeated words once each)
- repeated_words: a sorted list of words appearing more than once
- palindromes: a sorted list of palindrome words

- **Notes**:<br>
Treat words as case-insensitive (e.g., "Hello" and "hello" are the same word)
Remove any punctuation (.,!?:;"()) from words
Sort both the repeated_words and palindromes lists alphabetically

```python
def analyze_text(text):
    if not text:
        return {'unique_count': 0, 'repeated_words': [], 'palindromes': []}
    
    words=text.lower().split()

    cleaned_words=[]
    for w in words:
        clean_word = w.strip('.,!?:;"')
        if clean_word:
            cleaned_words.append(clean_word)
    word_count={}
    for word in cleaned_words:
        word_count[word]=word_count.get(word,0)+1
    
    unique_count=len(word_count)
    repeated_words=[word for word, count in word_count.items() if count > 1]
    repeated_words=sorted(repeated_words)

    palindromes=[word for word in word_count if word==word[::-1]]
    palindromes=sorted(palindromes)

    return {
        'unique_count': unique_count,
        'repeated_words': repeated_words,
        'palindromes': palindromes
    }
analyze_text('Wow! Did Hannah see that Race car? Mom was there too. Hannah did see it!')

#Lines 1-4: If nothing is given, return empty result.
#Step 1: Turn all text to small letters and break it into words.
#Step 2: Remove dots, commas, exclamation marks etc. from the sides of each word.
#Step 3: Count every word (e.g. "hello" appears 3 times → we remember 3).
#Step 4: unique_count = how many different words we have.
#Step 5: Collect only words that appeared more than 1 time, then sort them A→Z.
#Step 6: Check which words are palindromes (like "level", "radar"), then sort them A→Z.   
```
