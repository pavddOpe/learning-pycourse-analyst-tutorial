## Challenge

Write a function named merge that receives two lists as arguments and merges them into one sorted list.

**What you need to do**: Fill in the body of the function to:

Combine all elements from both lists into a single list<br>
Sort the combined list in ascending order<br>
Return the sorted list<br>
Example: merge([1, 4, 2], [2, 5, 9]) should return [1, 2, 2, 4, 5, 9]<br>

```python
def merge(lst1, lst2):
    combined_list = []
    for element in lst1:
        combined_list.append(element)
    for element in lst2:
        combined_list.append(element)
    combined_list.sort()
    return combined_list
merge([100,99,88,77],[1,9,2,8,3,7,5])
```

**Write a function** named prod which gets a list of numbers<br> as argument and returns the product of all the numbers in the list.<br>
Reminder, product is the multiplication of all the numbers, for [1, 2, 3], return 6 = 1 * 2 * 3.

```python
def prod(lst):
    var1 = 1
    for element in lst:
        var1 = var1*element
    return var1
prod([12,23,56,2,-18,98])
```

## **Challenge**

Write a function named reverse which gets a list of numbers as argument and returns the reversed list.

```python
def reverse(lst):
    rvrs_lst=[]
    n=len(lst)
    for i in range(n-1,-1,-1):     
        rvrs_lst.append(lst[i])
    return rvrs_lst
reverse([4,5,2,3])
```

## **Challenge**

*Create a function named combine_and_filter that receives 2 arguments:*

**First argument** is a list of numbers<br>
**Second argument** is an integer threshold value<br>
The function should:

Remove all numbers that are less than or equal to the threshold value (keep only numbers greater than the threshold)
Sort the remaining numbers in ascending order
Return the resulting list
For example, calling combine_and_filter([1, 5, 3, 2, 7, 4], 3) should return [4, 5, 7] because:

Numbers ≤ 3 are removed: 1, 3, 2
Numbers > 3 are kept: 5, 7, 4
After sorting: [4, 5, 7]

```python
def combine_and_filter(lst,threshold):
    combined_lst = []
    for i in range(len(lst)):
        number = lst[i]
        if number>threshold:
            combined_lst.append(number)
    combined_lst.sort()
    return combined_lst    
combine_and_filter([-1,10,5,-3,15,-2,8,0],0)
```

```python
def combine_and_filter(lst,threshold):
    return sorted(num for num in lst if num>threshold)
combine_and_filter([-1,10,5,-3,15,-2,8,0],0)
```
