## **Challenge**

Imagine you are managing an exclusive club where members are split into different groups based on their interests. <br> **You want to find out:**

- Who are the members that share a common interest in all groups?
- Who are the truly unique members who belong only to the first group and not to any others?<br>
**Write a program that:**

- Takes three sets, group1, group2, and group3, representing members of three interest groups.
- Finds and prints the intersection of all three groups (members common to all groups).
- Finds and prints the difference of group1 from the other two groups (members only in the first group).

```python
group1=eval(input())
group2=eval(input())
group3=eval(input())

intersection_result = group1 & group2 & group3
difference_result = group1 - group2 - group3

print("Members in all groups:", sorted(list(intersection_result)))
print("Unique members in group1:", sorted(list(difference_result)))
```

**Write a function generate_booth_directory** that takes visitors, health_topics and returns a dictionary mapping hashed visitor IDs to formatted booth assignments.

**The function creates unique booth** identifiers by combining visitor information with health topics, then generates a comprehensive directory for the wellness festival.<br>

- For each visitor, create a unique ID by hashing their name and age combined
- Assign health topics cyclically to visitors (first visitor gets first topic, etc.)
- Format booth information as: "BOOTH-{topic_upper} | Visitor: {name_title} (Age: {age}) | Topic: {topic_formatted}"
- Use SHA-256 hashing to generate 8-character alphanumeric IDs from visitor data<br>
**Parameters:**
- visitors (list): List of dictionaries with 'name' and 'age' keys
- health_topics (list): List of health topic strings
- Returns: Dictionary with hashed visitor IDs as keys and formatted booth strings as values. Format: {"abc12def": "BOOTH-MEDITATION | Visitor: John Smith (Age: 35) | Topic: Meditation & Mindfulness"}

```python
import hashlib

def generate_booth_directory(visitors,health_topics):
    final_dc={}
    for i, vis in enumerate(visitors):
        name = vis['name']
        age = vis['age']
        to_hash = name.lower() + str(age)
        hashed = hashlib.sha256(to_hash.encode()).hexdigest()[:8]
        topic = health_topics[i % len(health_topics)]
        booth_part = 'BOOTH'+'-'+''.join(topic.upper().replace(' ',''))
        visitor_part = f'Visitor: {name.title()} (Age: {age})'
        topic_part = f'Topic: {topic.title().replace(' ',' & ')}'
        fin_str = f'{booth_part} | {visitor_part} | {topic_part}'
        final_dc[hashed]=fin_str
    return final_dc
generate_booth_directory([
    {"name": "liam garcia", "age": 27},
    {"name": "Carol White", "age": 35}
],["aromatherapy", "herbal remedies", "meditation"])
```

**Create a function named filter_and_square_set** that takes a set input_set as an argument. The function should iterate over the elements in the set, filter out even numbers, and square the remaining odd numbers. The function should return a new set containing only the squared values of the odd numbers from the input set.

```python
def filter_and_square_set(input_set):
    new_set=set()
    for element in input_set:
        if element % 2 != 0:
            el1 = element**2
            new_set.add(el1)
    return new_set
filter_and_square_set({15,16,17,18,19})
```

**Create a function named add_student** that takes three arguments: name (string), age (integer), and courses (a list of strings). The function should:

```python
student_records={}
def add_student(name,age,courses):
    if name in student_records:
        print(f"Student '{name}' already exists.")
    else:
        student_records[name]={
            "age": age,
            "grades": set(),
            "courses": set(courses)
        }
        print(f"Student '{name}' added successfully.")
def add_grade(name,grade):
    if name in student_records:
        student_records[name]['grades']=grade
        print(f"Grade {grade} added for student '{name}'.")
    else:
        print(f"Student '{name}' not found.")

def is_enrolled(name,course):
    if name in student_records:
        if course in student_records[name]['courses']:
            return True
        else:
            return False
    else:
        print(f"Student '{name}' not found.")
        return False
add_student("Alice", 20, ["Math", "Physics"])
add_student("Bob", 22, ["Biology", "Chemistry"])
add_grade("Alice", 90)
add_grade("Alice", 85)
add_grade("Bob", 75)
add_grade("Charlie", 80)  # Non-existent student
print(is_enrolled("Alice", "Math"))  # Should return True
print(is_enrolled("Alice", "Biology"))  # Should return False
print(is_enrolled("Bob", "Biology"))  # Should return True
print(is_enrolled("Charlie", "Math"))  # Non-existent student, should print message and return False
```

**Create a function named moss_field_notes** that receives color, texture, and environment as its parameters.

The function should help a biology student, Alex, categorize and document moss species during their fieldwork expedition in a remote wilderness. Alex doesn't have access to modern technology, so they rely on simple observations and comparisons.

**The function should perform the following tasks:**<br>
- Concatenate the input parameters into a field note string in the format: "Color: [color], Texture: [texture], Environment: [environment]".<br>
- Compare the generated field note to a predefined list of known moss descriptions.
- Return a string that includes the field note and a tentative identification if a match is found, or a message indicating that the species is potentially new if no match is found.<br>

**Use the following list of known moss descriptions for comparison**:<br>
known_mosses = [
    "Color: Green, Texture: Soft, Environment: Shaded forest floor",
    "Color: Red, Texture: Rough, Environment: Exposed rocks",
    "Color: Brown, Texture: Fuzzy, Environment: Tree bark",
    "Color: Yellow, Texture: Spongy, Environment: Wetlands"
]

**Parameters:**
- color (str): The observed color of the moss.
- texture (str): The observed texture of the moss.
- environment (str): The environment where the moss was found.
- The function returns a string containing the field note and the result of the comparison.

```python
def moss_field_notes(color,texture,environment):
    field_note = (f"Color: {color}, Texture: {texture}, Environment: {environment}")
    known_mosses=[
        "Color: Green, Texture: Soft, Environment: Shaded forest floor",
        "Color: Red, Texture: Rough, Environment: Exposed rocks",
        "Color: Brown, Texture: Fuzzy, Environment: Tree bark",
        "Color: Yellow, Texture: Spongy, Environment: Wetlands"
    ]
    if field_note in known_mosses:
        idx = known_mosses.index(field_note)
        print(field_note)
        print(f'Tentative identification: Known moss species #{idx+1}')
    else:
        print(field_note)
        print('No match found. This could be a new species!')
moss_field_notes('Yellow','Spongy','Wetlands')
```
